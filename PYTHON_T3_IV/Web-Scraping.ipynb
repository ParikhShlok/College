## for one page with all 30 movie

````python
url='https://subslikescript.com/movies'
result=requests.get(url)
content=result.text
soup=BeautifulSoup(content,'html.parser')
print(soup.prettify())
````

````python
len(soup.find_all('a',href=True))
````

````python
box=soup.find('article',class_="main-article")
box.find_all('a',href=True)
````

````python
link=[]
for i in box.find_all('a',href=True):
    link.append(i['href'])
link
````

````python
url=[]
for i in link:
    url.append("https://subslikescript.com"+i)
url
````

````python
for i in url:
    result=requests.get(i)
    content=result.text
    soup=BeautifulSoup(content,'html.parser')
    box=soup.find('article','main-article')
    title=box.find('h1').text.strip()
    script=box.find('div',class_='full-script').text.strip()
    try:
        with open(f"{title}.txt",'w',encoding='utf-8') as f:
            f.write(script)
    except Exception as e:
        print(e)
    
````

````python
import requests
from bs4 import BeautifulSoup
````

````python
url="https://www.politifact.com/factchecks/"
result=requests.get(url)
content=result.text
soup=BeautifulSoup(content,'html.parser')
# print(soup.prettify())
````

````python
statement_all=[]
for i in soup.find_all('div',class_="m-statement__quote"):
    statement_all.append(i.find('a').text.strip())
statement_all
````

````python
source_all=[]
for i in soup.find_all('div',class_="m-statement__meta"):
    source_all.append(i.find('a').text.strip())
source_all
````

````python
date=soup.find('footer',class_="m-statement__footer").text.strip().split('• ')[-1]
date
date_all=[]
for i in soup.find_all('footer',class_="m-statement__footer"):
    date_all.append(i.text.strip().split('• ')[-1])
date_all
````

````python
link=soup.find("div",class_='m-statement__quote').find('a')['href']
link_all=[]
for i in soup.find_all("div",class_='m-statement__quote'):
    link_all.append("https://www.politifact.com"+i.find('a')['href'])
link_all
````

# WITH FILE

````python
from bs4 import BeautifulSoup
````

````python
soup=BeautifulSoup(open('top250_2026.html','r',encoding='utf-8'))
print(soup.prettify())
````

````python
title_all=[]
for i in soup.find_all('td',class_="titleColumn"):
    title_all.append(i.find('a').text.strip())
title_all
````

````python
year_all=[]
for i in soup.find_all('span',class_="secondaryInfo"):
    year_all.append(i.text.strip()[1:-1])
year_all
````

````python
rating_all=[]
for i in soup.find_all('td',class_="ratingColumn imdbRating"):
    rating_all.append(i.text.strip())
rating_all
````

````python
import pandas as pd
````

````python
df=pd.DataFrame({
    "Movie Name":title_all,
    "year":year_all,
    "rating":rating_all
})
df
````
